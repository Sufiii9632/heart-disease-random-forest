```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score, classification_report

FILENAME = 'heart.xlsx'
TARGET_COL = 'target'

xls = pd.ExcelFile(FILENAME)
df = pd.read_excel(FILENAME, sheet_name=xls.sheet_names[0])
df.columns = df.columns.str.strip()
df = df.dropna(subset=[TARGET_COL])

print("Dataset Shape:", df.shape)
display(df.head())

y = df[TARGET_COL]
X = df.drop(columns=[TARGET_COL])
feature_names = X.columns

X_imputed = SimpleImputer(strategy='median').fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X_imputed, y, test_size=0.2, random_state=42, stratify=y)

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("\n--- Random Forest Evaluation Metrics ---")
print(f"Accuracy : {acc*100:.2f}%")
print(f"Precision: {prec*100:.2f}%")
print(f"Recall   : {rec*100:.2f}%")
print(f"F1-Score : {f1*100:.2f}%")
print("\nClassification Report:\n", classification_report(y_test, y_pred))

plt.style.use('dark_background')
fig, axes = plt.subplots(1, 3, figsize=(17, 5.2))
fig.patch.set_facecolor('#0F172A')
plt.setp(axes, facecolor='#1E293B')
_ = [sp.set_color('#334155') for ax in axes for sp in ax.spines.values()]

cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='mako', cbar=False, ax=axes[0], annot_kws={'size': 14, 'weight': 'bold', 'color': 'white'}, xticklabels=['Pred: Healthy (0)', 'Pred: Disease (1)'], yticklabels=['Actual: Healthy (0)', 'Actual: Disease (1)'])
axes[0].set_title('Confusion Matrix', fontsize=13, fontweight='bold', color='#F8FAFC', pad=12)

metrics_names = ['Accuracy', 'Precision', 'Recall', 'F1-Score']
metrics_values = [acc, prec, rec, f1]
bars = axes[1].bar(metrics_names, metrics_values, color=['#38BDF8', '#818CF8', '#E879F9', '#34D399'], width=0.55)
axes[1].set_ylim(0, 1.15)
axes[1].set_title('Model Performance Metrics', fontsize=13, fontweight='bold', color='#F8FAFC', pad=12)
axes[1].grid(axis='y', linestyle='--', alpha=0.2)
axes[1].bar_label(bars, labels=[f'{v*100:.1f}%' for v in metrics_values], padding=4, color='#F8FAFC', fontweight='bold', fontsize=10.5)

importances = pd.Series(model.feature_importances_, index=feature_names).nlargest(6).sort_values()
axes[2].barh(importances.index, importances.values, color='#38BDF8', height=0.55)
axes[2].set_title('Top 6 Predictive Features', fontsize=13, fontweight='bold', color='#F8FAFC', pad=12)
axes[2].set_xlabel('Importance Score', color='#94A3B8')
axes[2].grid(axis='x', linestyle='--', alpha=0.2)

plt.tight_layout()
plt.savefig('evaluation_results.png', dpi=300, facecolor=fig.get_facecolor())
plt.show()
```
